```python
#activate pytorch_amd
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from diffusers import UNet2DModel
import matplotlib.pyplot as plt
from torch.utils.data import Subset
import torch.nn.functional as F
import torch_directml
```

```python
device = torch_directml.device(1)
print(f"Using device: {device}")
```

```python
sigma_q = 1
batch_size = 256
transform_pipeline = transforms.Compose([
    transforms.ToTensor(),               
    transforms.Normalize((0.5,), (0.5,)) #rescale to [-1.0, 1.0]
])
train_dataset = torchvision.datasets.MNIST(
    root='./data', 
    train=True,        
    download=True,       
    transform=transform_pipeline 
)

train_loader = torch.utils.data.DataLoader(
    train_dataset, 
    batch_size=batch_size,       
    shuffle=True,
    num_workers=4,        
    pin_memory=True
)

# take 1000 subset to test 
mini_dataset = Subset(train_dataset, range(1000))

# test loader
mini_train_loader = torch.utils.data.DataLoader(
    mini_dataset,
    batch_size=batch_size,
    shuffle=True
)
```

```python
# pic extraction
images, labels = next(iter(train_loader))
```

```python
fig, axes = plt.subplots(4, 4, figsize=(4, 4))

for i, ax in enumerate(axes.flatten()):
    img = torch.clamp(images[6*i], 0.0, 1.0).permute(1, 2, 0).cpu().numpy()
    
    ax.imshow(img, cmap='gray')
    ax.axis('off') 

plt.tight_layout()
plt.show()
```

```python
#U-Net untrained (for v1 use 32,64,64 and sigma_q=1, for others - 64,128,256)
model = UNet2DModel(
    sample_size=28,  
    in_channels=1,   
    out_channels=1,  
    layers_per_block=2, 
    block_out_channels=(64,128,256), 
    down_block_types=("DownBlock2D", "AttnDownBlock2D", "DownBlock2D"),
    up_block_types=("UpBlock2D", "AttnUpBlock2D", "UpBlock2D")
)
```

```python
optimizer = optim.Adam(model.parameters(), lr=1e-4)
epochs = 15 

model = model.to(device)
first_weight_device = next(model.parameters()).device
print(f"The U-Net is currently sitting on: {first_weight_device}")

model.train() 

for epoch in range(epochs):
    total_loss = 0
    
    for batch_images, _ in train_loader:
        
        optimizer.zero_grad()
        batch_images = batch_images.to(device) 
        t_math = torch.rand(batch_images.shape[0], 1, 1, 1, device=device)
        noise = torch.randn_like(batch_images)
        # Calculate your Brownian standard deviation
        std = torch.sqrt(t_math) * sigma_q
        # Create the massive noisy image
        x_t = batch_images + noise * std

        noise_pred = model(x_t, t_math.squeeze()).sample
        loss = F.mse_loss(noise_pred, noise*std )
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
        

    avg_loss = total_loss / len(train_loader)
    print(f"Epoch {epoch+1} | Average Loss: {avg_loss:.4f}")
```

```python
#for pretrained models
ckpt = torch.load("mnist_diffusion_v40.pt", map_location=device)
model.load_state_dict(ckpt)
model = model.to(device)
model.eval()
```

### DDPM Reverse Sampling

```python
model.eval()
size = 1
sigma_q = 1
x = torch.randn((size, 1, 28, 28)).to(device) * sigma_q 
dt = 0.001
t_current = 1.0

print("Starting reverse diffusion...")

with torch.no_grad():
    while t_current > 0:

        t_model_input = torch.full((size,), t_current*1000).to(device)
        
        eps = model(x, t_model_input).sample
        
        t_tensor = torch.full((size, 1, 1, 1), t_current)

        mu = x - dt/t_current*eps
        x = mu + torch.randn((size,1,28,28)).to(device)*sigma_q*np.sqrt(dt)
        t_current -= dt


print("Generation complete! Final shape:", x.shape)
```
